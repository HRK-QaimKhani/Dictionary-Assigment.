```python
# Dictionary Assigment:
# Dictionary definition:
# A dictionary in Python is a collection that stores data in key–value pairs,
# where each key is used to find its value—like a real dictionary where a word (key) has a meaning (value).
```

```python
# Create a Dictionary:
# In python,a dictionary can be created by placing a sequence of elements within curly{} braces,seprated by'comma'.
dict={'brand':'Ford','model':'mustang','year':1984}
print(dict)
```

```python
# Dictionary Item:
# A dictionary item in Python is one key–value pair inside a dictionary, where a key is linked to its corresponding value.
dict={"1":"car","2":"bike","3":"train","4":"ship","5":"hello","6":"HRk"}
print(dict['1'])
print(dict['4'])
```

```python
# Dictionary Length:
# Dictionary length in Python is the number of key–value pairs in a dictionary, found using the len() function.
dict={"1":"car","2":"bike","3":"train","4":"ship","5":"hello","6":"HRk"}
print(len(dict))

# type()
#Type in Python tells you what kind of data a value or variable is,
# such as number, string, list, or dictionary, and is checked using the type() function.
print(type(dict))
```

```python
# Acces
# Access in Python means getting or taking a value from something, like taking a value from a dictionary using its key.
dict={"1":"car","2":"bike","3":"train","4":"ship","5":"hello","6":"HRk"}
x=dict['5']
print(x)

# Another method called to acces is get() that will give you the same result.
y=dict.get('2')
print(y)
```

```python
# Get Keys:
# Get keys in Python means finding or listing all the keys in a dictionary.
dict={"1":"car","2":"bike","3":"train","4":"ship","5":"hello","6":"HRk"}
x=dict.keys()
print(x)

dict['7']='alto'
print(dict)

print(x)
```

```python
# Get items:
# Get items in Python means getting all the key–value pairs from a dictionary.
dict={"1":"car","2":"bike","3":"train","4":"ship","5":"hello","6":"HRk"}
b=dict.items()
print(b)

dict['year']=1948
print(b)
```

```python
# Check existence of Key:
# To determine the specified key is present in a dictionary use the in keyword.
dict={"1":"car","2":"bike","3":"train","4":"ship","5":"hello","6":"HRk"}
if '8' in dict:
  print("yes,it is one of the keys in the dictionary")
else:
  print("not in the list")
```

```python
# Change and Update:
# Change and Update in Python means modifying an existing value in a dictionary or variable,
# or adding new key–value pairs to a dictionary.
dict={"1":"car","2":"bike","3":"train","4":"ship","5":"hello","6":"HRk"}
dict['1']='apple'
print(dict)

dict.update({ '1' : ''})
```

```python
# Remove Items:
# POP >> Remove item using pop() in Python means deleting a specific key–value pair from a dictionary by giving its key.
dict={"1":"car","2":"bike","3":"train","4":"ship","5":"hello","name":"HRk"}
dict.pop('name')

# POPItem >> Remove item using popitem() in Python means deleting the last added key–value pair from a dictionary.
dict.popitem()
# del >> Remove item using del in Python means deleting a key–value pair from a dictionary or removing a variable completely.
del dict['1']
print(dict)

# clear >> clear() in Python means removing all items from a dictionary, leaving it empty.
dict.clear()
print(dict)
```

```python
# Nested Dictionary:
# Nested dictionary in Python is a dictionary inside another dictionary, like a dictionary within a dictionary.
people = {
    "person1": {
        "name": "Alice",
        "age": 30,
        "contact": {
            "email": "alice@example.com",
            "phone": "123-456-7890"
        },
        "hobbies": ["reading", "traveling"]
    },
    "person2": {
        "name": "Bob",
        "age": 25,
        "contact": {
            "email": "bob@example.com",
            "phone": "987-654-3210"
        },
        "hobbies": ["sports", "gaming"]
    },
    "person3": {
        "name": "Charlie",
        "age": 28,
        "contact": {
            "email": "charlie@example.com",
            "phone": "555-555-5555"
        },
        "hobbies": ["music", "photography"]
    }
}
print(people)
```
